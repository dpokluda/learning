```python
from torchvision import datasets, transforms

train_set = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transforms.ToTensor(),
)
```

```python
import matplotlib.pyplot as plt
import torch

# Pick 25 random indices from the training set
random_indices = torch.randperm(len(train_set))[:25]

%matplotlib inline
fig, axes = plt.subplots(5, 5, figsize=(8, 8))
for ax, idx in zip(axes.flat, random_indices):
    image, label = train_set[idx]   # image: [1, 28, 28] tensor, label: int
    ax.imshow(image.squeeze(), cmap="gray")
    ax.set_title(str(label))
    ax.axis("off")

plt.tight_layout()
```

```python
# create tenzor for our batch of images from random_indices
batch_images = torch.stack([train_set[idx][0] for idx in random_indices])
print(batch_images.shape)  
```

```python
# check the min and max pixel values in the batch
print(f"min={batch_images.min().item()}, max={batch_images.max().item()}")
```

```python
# use DataLoader to create batches of data
from torch.utils.data import DataLoader
batch_size = 25
train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True)
# get one batch of data
batch_images, batch_labels = next(iter(train_loader))
print(batch_images.shape)  # should be [25, 1, 28, 28]
print(batch_labels.shape)  # should be [25]
print(f"min={batch_images.min().item()}, max={batch_images.max().item()}")
```

```python
fig, axes = plt.subplots(5, 5, figsize=(8, 8))
for ax, image, label in zip(axes.flat, batch_images, batch_labels):
    ax.imshow(image.squeeze(), cmap="gray")
    ax.set_title(str(label.item()))
    ax.axis("off")

plt.tight_layout()
```
